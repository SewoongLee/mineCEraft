# Function: `are_adjacent`

```python
from eval_code.shape import are_adjacent

# adjacent: door + sea_lantern (substring_match)
coords = [
    {"x": 0, "y": 0, "z": 0, "material": "oak_door"},
    {"x": 1, "y": 0, "z": 0, "material": "sea_lantern"},
]
assert True == are_adjacent(coords, ["door", "sea_lantern"], substring_match=True)
plotly_blocks.plot(coords, title="are_adjacent: door + sea_lantern (adjacent)")

# not adjacent
coords = [
    {"x": 0, "y": 0, "z": 0, "material": "oak_door"},
    {"x": 5, "y": 0, "z": 0, "material": "sea_lantern"},
]
assert False == are_adjacent(coords, ["door", "sea_lantern"])
plotly_blocks.plot(coords, title="are_adjacent: door + sea_lantern (not adjacent)")

# missing one material -> False
coords = [{"x": 0, "y": 0, "z": 0, "material": "sea_lantern"}]
assert False == are_adjacent(coords, ["door", "sea_lantern"])
plotly_blocks.plot(coords, title="are_adjacent: missing door")
```

# Function: `are_doors_passable`

```python
from eval_code.shape import are_doors_passable

coords = [
    {"x":  0, "y": 1, "z": 0, "material": "stone"},
    {"x":  1, "y": 1, "z": 0, "material": "oak_door"},
    {"x":  2, "y": 1, "z": 0, "material": "stone"},
]
assert True == are_doors_passable(coords, verbose=True)

coords = [
    {"x":  0, "y": 1, "z": 0, "material": "stone"},
    {"x":  0, "y": 1, "z": 1, "material": "oak_door"},
    {"x":  0, "y": 1, "z": 2, "material": "stone"},
]
assert True == are_doors_passable(coords, verbose=True)

coords = [
    {"x":  0, "y": 1, "z": 0, "material": "stone"},
    {"x":  0, "y": 1, "z": 1, "material": "oak_door"}, # Not passable
    {"x":  0, "y": 1, "z": 2, "material": "stone"},
    {"x":  1, "y": 1, "z": 1, "material": "stone"},
]
assert False == are_doors_passable(coords, verbose=True)

coords = [
    {"x":  0, "y": 1, "z": 0, "material": "stone"},
    {"x":  1, "y": 0, "z": 0, "material": "oak_door"},
    {"x":  2, "y": 1, "z": 0, "material": "stone"},
]
assert True == are_doors_passable(coords, verbose=True) # Doors are 2 blocks tall, so both y and y+1 levels are checked.
```

# Accessibility: `is_single_level`, `has_exit`, `has_wide_exit`

- **is_single_level**: bed-based, flat path on ground (y=min) to outside. Bed must be on ground.
- **has_exit** / **has_wide_exit**: centroid-based (or start_material), BFS with ±1 y, doors passable.

```python
from eval_code.shape import is_single_level, has_exit, has_wide_exit


def _build(floor_y: int, exit_width: int):
    """
    Build 5x5 room. floor_y=-1: ground; floor_y=0: raised floor.
    exit_width: 0=closed, 1=1-block corridor (1 oak_door), 2=2-block gap (2 oak_doors).
    Doors at opening to verify they are treated as passable.
    """
    bed_y = floor_y + 1
    coords = []
    for x in range(-2, 3):
        for z in range(-2, 3):
            coords.append({"x": x, "y": floor_y, "z": z, "material": "dirt" if floor_y == -1 else "oak_planks"})
    coords.append({"x": 0, "y": bed_y, "z": 0, "material": "oak_bed"})
    wall_y_start = floor_y + 1
    wall_heights = 3 if floor_y == -1 else 2
    for x in range(-2, 3):
        for z in range(-2, 3):
            if x in (-2, 2) or z in (-2, 2):
                if exit_width == 1:
                    skip = z in (-2, -1) and x == 0
                elif exit_width == 2:
                    skip = z in (-2, -1) and x in (-1, 0)
                else:
                    skip = False
                if not skip:
                    for i in range(wall_heights):
                        coords.append({"x": x, "y": wall_y_start + i, "z": z, "material": "stone"})
    if exit_width >= 1:
        for i in range(exit_width):
            xx = -1 + i if exit_width == 2 else 0
            coords.append({"x": xx, "y": wall_y_start, "z": -2, "material": "oak_door"})
    return coords


# --- Case A: floor_y=-1 (ground), is_single_level PASS when exit exists ---
# 1. Closed: single-level✗ exit✗ wide✗
coords = _build(floor_y=-1, exit_width=0)
plotly_blocks.plot(coords, title="A.1 closed: single-level✗ exit✗ wide✗")
assert False == is_single_level(coords, start_material="bed")
assert False == has_exit(coords)
assert False == has_wide_exit(coords)

# 2. 1-block: single-level✓ exit✓ wide✗
coords = _build(floor_y=-1, exit_width=1)
plotly_blocks.plot(coords, title="A.2 1-block: single-level✓ exit✓ wide✗")
assert True == is_single_level(coords, start_material="bed")
assert True == has_exit(coords)
assert False == has_wide_exit(coords)

# 3. 2-block: single-level✓ exit✓ wide✓
coords = _build(floor_y=-1, exit_width=2)
plotly_blocks.plot(coords, title="A.3 2-block: single-level✓ exit✓ wide✓")
assert True == is_single_level(coords, start_material="bed")
assert True == has_exit(coords)
assert True == has_wide_exit(coords)

# --- Case B: floor_y=0 (raised), is_single_level always FAIL ---
# 1. Closed: single-level✗ exit✗ wide✗
coords = _build(floor_y=0, exit_width=0)
plotly_blocks.plot(coords, title="B.1 closed: single-level✗ exit✗ wide✗")
assert False == is_single_level(coords, start_material="bed")
assert False == has_exit(coords)
assert False == has_wide_exit(coords)

# 2. 1-block: single-level✗ exit✓ wide✗
coords = _build(floor_y=0, exit_width=1)
plotly_blocks.plot(coords, title="B.2 1-block: single-level✗ exit✓ wide✗")
assert False == is_single_level(coords, start_material="bed")
assert True == has_exit(coords)
assert False == has_wide_exit(coords)

# 3. 2-block: single-level✗ exit✓ wide✓
coords = _build(floor_y=0, exit_width=2)
plotly_blocks.plot(coords, title="B.3 2-block: single-level✗ exit✓ wide✓")
assert False == is_single_level(coords, start_material="bed")
assert True == has_exit(coords)
assert True == has_wide_exit(coords)

print("All accessibility tests passed.")
```
