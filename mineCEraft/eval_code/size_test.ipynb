# Function: `is_equal`

```python
from eval_code.size import is_equal

# Test case: 3x3x2 structure
coords = [
    # Bottom layer (y=0)
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 1, "y": 0, "z": 0, "material": "stone"},
    {"x": 2, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 0, "z": 1, "material": "stone"},
    {"x": 1, "y": 0, "z": 1, "material": "stone"},
    {"x": 2, "y": 0, "z": 1, "material": "stone"},
    {"x": 0, "y": 0, "z": 2, "material": "stone"},
    {"x": 1, "y": 0, "z": 2, "material": "stone"},
    {"x": 2, "y": 0, "z": 2, "material": "stone"},
    # Top layer (y=1)
    {"x": 0, "y": 1, "z": 0, "material": "stone"},
    {"x": 1, "y": 1, "z": 0, "material": "stone"},
    {"x": 2, "y": 1, "z": 0, "material": "stone"},
    {"x": 0, "y": 1, "z": 1, "material": "stone"},
    {"x": 1, "y": 1, "z": 1, "material": "stone"},
    {"x": 2, "y": 1, "z": 1, "material": "stone"},
    {"x": 0, "y": 1, "z": 2, "material": "stone"},
    {"x": 1, "y": 1, "z": 2, "material": "stone"},
    {"x": 2, "y": 1, "z": 2, "material": "stone"},
]

# Expect True: 3x3 in xz, 2 in y (range: dx=2, dz=2, dy=1)
assert True == is_equal(coords, xz=(3, 3), y=2)
print("✓ PASS: is_equal(coords, xz=(3, 3), y=2)")

# Expect False: Wrong size (should be 3x3x2, not 4x4x2)
assert False == is_equal(coords, xz=(4, 4), y=2)
print("✓ PASS: is_equal(coords, xz=(4, 4), y=2) returns 0")

# Expect False: Wrong height (should be 2, not 3)
assert False == is_equal(coords, xz=(3, 3), y=3)
print("✓ PASS: is_equal(coords, xz=(3, 3), y=3) returns 0")

plotly_blocks.plot(coords)
```

# Function: `is_leq`

```python
from eval_code.size import is_leq

# Test case: 2x2x2 structure (smaller than 3x3x3)
coords = [
    # Bottom layer (y=0)
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 1, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 0, "z": 1, "material": "stone"},
    {"x": 1, "y": 0, "z": 1, "material": "stone"},
    # Top layer (y=1)
    {"x": 0, "y": 1, "z": 0, "material": "stone"},
    {"x": 1, "y": 1, "z": 0, "material": "stone"},
    {"x": 0, "y": 1, "z": 1, "material": "stone"},
    {"x": 1, "y": 1, "z": 1, "material": "stone"},
]

# Expect True: 2x2x2 is within bounds of 3x3x3
assert True == is_leq(coords, xz=(3, 3), y=3)
print("✓ PASS: is_leq(coords, xz=(3, 3), y=3)")

# Expect True: 2x2x2 is within bounds of 2x2x2
assert True == is_leq(coords, xz=(2, 2), y=2)
print("✓ PASS: is_leq(coords, xz=(2, 2), y=2)")

# Expect False: 2x2x2 exceeds bounds of 1x1x1 in xz
assert False == is_leq(coords, xz=(1, 1), y=1)
print("✓ PASS: is_leq(coords, xz=(1, 1), y=1) returns 0")

# Expect False: 2x2x2 exceeds bounds in y (y=1 means max range is 0, but we have range 1)
assert False == is_leq(coords, xz=(3, 3), y=1)
print("✓ PASS: is_leq(coords, xz=(3, 3), y=1) returns 0")

# Expect False: xz is OK but y exceeds
assert False == is_leq(coords, xz=(3, 3), y=0)
print("✓ PASS: is_leq(coords, xz=(3, 3), y=0) returns 0")

plotly_blocks.plot(coords)
```
