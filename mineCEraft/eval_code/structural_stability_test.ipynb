```python
from eval_code.stress import compute_von_mises_stress

coords = [
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 1, "z": 0, "material": "stone"},
    {"x": 1, "y": 1, "z": 0, "material": "stone"},
    {"x": 2, "y": 1, "z": 0, "material": "stone"},
    {"x": 3, "y": 1, "z": 0, "material": "stone"},
    {"x": 3, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 0, "z": 1, "material": "stone"},
    {"x": 0, "y": 1, "z": 1, "material": "stone"},
    {"x": 1, "y": 1, "z": 1, "material": "stone"},
    {"x": 2, "y": 1, "z": 1, "material": "stone"},
    {"x": 3, "y": 1, "z": 1, "material": "stone"},
    {"x": 3, "y": 0, "z": 1, "material": "stone"},
]

res = compute_von_mises_stress(coords, h_threshold=2.0, eta=0.0)
print(res.von_mises)  # (n,) Pa

plotly_blocks.plot(coords, color=res.von_mises, colorbar_title="von Mises Stress")
```

```python
coords = []

for xi in range(4):
    for zi in range(1, 3):
        for yi in range(7):
            coords.append({"x": xi, "y": yi, "z": zi, "material": "stone"})

for xi in range(4):
    for zi in range(3, 26):
        for yi in range(5, 7):
            coords.append({"x": xi, "y": yi, "z": zi, "material": "stone"})

for xi in range(4):
    for zi in range(26, 28):
        for yi in range(7):
            coords.append({"x": xi, "y": yi, "z": zi, "material": "stone"})

coords.append({"x": 0, "y": 4, "z": 3, "material": "stone"})
coords.append({"x": 1, "y": 4, "z": 3, "material": "stone"})
coords.append({"x": 2, "y": 4, "z": 3, "material": "stone"})
coords.append({"x": 3, "y": 4, "z": 3, "material": "stone"})
coords.append({"x": 0, "y": 4, "z": 25, "material": "stone"})
coords.append({"x": 1, "y": 4, "z": 25, "material": "stone"})
coords.append({"x": 2, "y": 4, "z": 25, "material": "stone"})
coords.append({"x": 3, "y": 4, "z": 25, "material": "stone"})
```
