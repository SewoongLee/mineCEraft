# Structure: Roof

```python
from eval_code.utils.stress import compute_von_mises_stress
from eval_code.structural_stability import stress_score

# Prompt: "Build a 10 × 10 flat roof supported by exactly five 2 x 2 x 2 columns. Arrange the columns so that the roof is as structurally stable as possible."
# Five columns
column_bases = [(4, 4), (1, 1), (1, 7), (7, 1), (7, 7)]
coords = []
for (cx, cz) in column_bases:
    for y in range(2):
        for dx in range(2):
            for dz in range(2):
                coords.append({'x': cx + dx, 'y': y, 'z': cz + dz})
# Roof: 10x10 slab layer at y=2
for x in range(10):
    for z in range(10):
        coords.append({'x': x, 'y': 2, 'z': z})

# Continuous score in [0, 1]: reference_stress / max_stress, capped at 1.0
assert stress_score(coords, 62e6) >= 1.0  # safe
assert 0 <= stress_score(coords, 61e6) < 1.0  # over threshold

# Calling intermediate functions again for visualization
von_mises = compute_von_mises_stress(coords)
print('Max von Mises Stress (Pa):', max(von_mises))
plotly_blocks.plot(coords, color=von_mises, colorbar_title="von Mises Stress (Pa)", alpha=1)
```

# Errors for Singular Structures

```python
# Caution!
# When calculating the stress, no part of the structure should be line (e.g., 1x1x3 pillars).
# This is because the singularity in calculating the von Mises stress will throw errors.
# Currently, all structural stability evaluation prompts are designed to prevent singularities.
# It is perfectly normal to encounter the following error during evaluation if agents construct singular structures.

from eval_code.utils.stress import compute_von_mises_stress
from eval_code.structural_stability import stress_score

# Singular structure => stress cannot be computed; score is 0.0
coords = [
 {'x': 0, 'y': 0, 'z': 0, 'material': 'stone_slab'},
 {'x': 0, 'y': 1, 'z': 0, 'material': 'stone_slab'},
 {'x': 0, 'y': 2, 'z': 0, 'material': 'stone_slab'}]

assert stress_score(coords, 1e6) == 0.0  # invalid/singular structure => 0 score

# Optionally visualize (may raise or produce NaNs)
try:
    von_mises = compute_von_mises_stress(coords)
    print("Max von Mises (if finite):", max(von_mises))
    plotly_blocks.plot(coords, color=von_mises, colorbar_title="von Mises Stress (Pa)", alpha=1)
except Exception as e:
    print("compute_von_mises_stress failed as expected:", e)
```
