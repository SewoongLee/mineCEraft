# `are_all_blocks_supported`

```python
from eval_code.structural_stability import are_all_blocks_supported

# ---------------------------------------------------------------------------
# Stable example
# ---------------------------------------------------------------------------
# Vertical support column at x = 0 (y = 0..1), and a horizontal beam along +x.
# Overhang length from x=0 to x=9 -> span = 9, which is within stone's limit.

stable_coords = [
    # vertical support column (2 blocks high)
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 1, "z": 0, "material": "stone"},
]

# horizontal beam from x = 1 to x = 9 at y = 1
stable_coords += [
    {"x": x, "y": 1, "z": 0, "material": "stone"}
    for x in range(1, 10)  # 1..9 inclusive
]

assert True == are_all_blocks_supported(stable_coords)
plotly_blocks.plot(stable_coords)
```

```python
# ---------------------------------------------------------------------------
# Unstable example
# ---------------------------------------------------------------------------
# Same idea as above, but the beam extends one block further (x = 10).
# Overhang length from x=0 to x=10 -> span = 10 (> stone limit 9).

unstable_coords = [
    # vertical support column (2 blocks high)
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 1, "z": 0, "material": "stone"},
]

# horizontal beam from x = 1 to x = 10 at y = 1
unstable_coords += [
    {"x": x, "y": 1, "z": 0, "material": "stone"}
    for x in range(1, 11)  # 1..10 inclusive
]

assert False == are_all_blocks_supported(unstable_coords)
plotly_blocks.plot(unstable_coords)
```
