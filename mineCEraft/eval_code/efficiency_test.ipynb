# `l1_dist`

```python
from eval_code.efficiency import l1_dist

coords = [
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 0, "y": 0, "z": 1, "material": "stone"},
    {"x": 1, "y": 0, "z": 0, "material": "stone"},
    {"x": 1, "y": 0, "z": 1, "material": "stone"},
]

# Order:
# 1 2
# 3 4
# => NOT efficient! (when placing from 2 to 3: step 2->3 has L1=2)
# L1_total = 1+2+1 = 4, optimal = 3 => ratio 3/4 = 0.75
assert 0 < l1_dist(coords) < 1
assert abs(l1_dist(coords) - 0.75) < 1e-6
```

```python
from eval_code.efficiency import l1_dist

coords = [
    {"x": 0, "y": 0, "z": 0, "material": "stone"},
    {"x": 1, "y": 0, "z": 0, "material": "stone"},
    {"x": 1, "y": 0, "z": 1, "material": "stone"},
    {"x": 0, "y": 0, "z": 1, "material": "stone"},
]

# Order:
# 1 2
# 4 3
# => efficient! (each step L1=1, ratio = 1.0)
assert l1_dist(coords) == 1.0
```
