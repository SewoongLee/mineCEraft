```python
from action_processor import read_coords_from_action
from eval_code import material, size, planning

# 1) Choose the JS file you want to evaluate
file_name = "0_rect.js"  # e.g., your rectangular foundation action

# 2) Extract coords from the JS action
coords = read_coords_from_action(file_name)

# 3) Build scores
score = {"material": 0, "size": 0, "planning": 0}
total = {"material": 0, "size": 0, "planning": 0}

# Material checks
total["material"] += 1; score["material"] += material.is_quantity_correct(coords, 15 * 20 * 2)  # 15x20x2 blocks
total["material"] += 1; score["material"] += material.is_type_correct(coords, "stone")

# Size check: xz=(15,20), y=2 layers
total["size"] += 1; score["size"] += size.is_equal(coords, xz=(15, 20), y=2)

# Planning (path efficiency): every consecutive step must be adjacent (L1=1)
total["planning"] += 1; score["planning"] += planning.l1_dist(coords, 1)


print("score:", score)
print("total:", total)
```
