# Total problem number (benchmarks + archive)
Counts problems from both `benchmarks/*.json` and `benchmarks/archive/*.json`.

```python
from pathlib import Path
import json

# Load JSON from benchmarks/ and benchmarks/archive/ (archive included)
# Schema: prompts = [[turn1, turn2, ...], ...], checks = [[eval_turn1], [eval_turn2], ...]
BENCHMARKS_DIR = Path.cwd() / "benchmarks"
json_files = sorted(BENCHMARKS_DIR.glob("*.json")) + sorted(BENCHMARKS_DIR.glob("archive/*.json"))
mapping = []
for json_file in json_files:
    file_mapping = json.loads(json_file.read_text(encoding="utf-8"))
    mapping.extend(file_mapping)

# Build runs: each run = (prompt_sequence, checks_per_turn)
runs = []
for item in mapping:
    for prompt_sequence in item["prompts"]:
        runs.append((prompt_sequence, item["checks"]))

print(f"[PY] Benchmarks dir: {BENCHMARKS_DIR.resolve()} (cwd: {Path.cwd().resolve()})")
print(f"Total problem # (including archive): {len(runs)}")
if len(runs) == 0:
    print("[PY] No runs. Run notebook from the folder that contains benchmarks/.")

for i, (prompts, _) in enumerate(runs):
    print(i, prompts)
```

```python
# Per-file problem counts (benchmarks + archive)
json_files = sorted(BENCHMARKS_DIR.glob("*.json")) + sorted(BENCHMARKS_DIR.glob("archive/*.json"))
for json_file in json_files:
    file_mapping = json.loads(json_file.read_text(encoding="utf-8"))
    count = sum(len(item["prompts"]) for item in file_mapping)
    rel = json_file.relative_to(BENCHMARKS_DIR)
    print(f"{rel}: {count}")
```
