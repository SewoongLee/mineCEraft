# MineCEraft Lite problem set
Counts problems using the same sampling logic as `main.ipynb` when `sampling_for_lite=True`.
Each JSON file in `benchmarks/` contributes one randomly sampled run. Same `sampling_rand_seed` yields the same problem set.

```python
sampling_rand_seed = 42  # Same seed as main.ipynb → same problem set
```

```python
from pathlib import Path
import json
import random

# Same logic as main.ipynb when sampling_for_lite=True
# Load JSON from benchmarks/ only (archive excluded, same as main)
# Schema: prompts = [[turn1, turn2, ...], ...], checks = [[eval_turn1], [eval_turn2], ...]
BENCHMARKS_DIR = Path.cwd() / "benchmarks"
json_files = sorted(BENCHMARKS_DIR.glob("*.json"))

rng = random.Random(sampling_rand_seed)
runs = []
for json_file in json_files:
    file_mapping = json.loads(json_file.read_text(encoding="utf-8"))
    file_runs = []
    for item in file_mapping:
        for prompt_sequence in item["prompts"]:
            file_runs.append((prompt_sequence, item["checks"], item.get("_comment", "")))
    if file_runs:
        runs.append(rng.choice(file_runs))

print(f"[PY] Benchmarks dir: {BENCHMARKS_DIR.resolve()} (cwd: {Path.cwd().resolve()})")
print(f"Total problem # (lite, seed={sampling_rand_seed}): {len(runs)}")
if len(runs) == 0:
    print("[PY] No runs. Run notebook from the folder that contains benchmarks/.")

for i, (prompts, _, _) in enumerate(runs):
    print(i, prompts)
```

```python
# Per-file: which run was sampled (lite)
rng = random.Random(sampling_rand_seed)
for json_file in json_files:
    file_mapping = json.loads(json_file.read_text(encoding="utf-8"))
    file_runs = []
    for item in file_mapping:
        for prompt_sequence in item["prompts"]:
            file_runs.append((prompt_sequence, item["checks"], item.get("_comment", "")))
    if file_runs:
        chosen = rng.choice(file_runs)
        rel = json_file.relative_to(BENCHMARKS_DIR)
        print(f"{rel}: {chosen[0]}")
```
